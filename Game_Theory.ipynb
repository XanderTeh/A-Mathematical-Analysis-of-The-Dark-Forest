# 1. Introduction
Game theory is the field of mathematics that studies and models interactions between multiple players in _games_ of strategy. Game theory describes, predicts, and evaluates the decisions players make when their outcomes depend not only on their own choices, but also on the choices of others. 

In game theory, a game is a mathematical model of a strategic interaction. A game consists of players, as well as the choices or strategies available to them, and the outcomes associated with different combinations of choices. These outcomes are commonly represented using payoffs, which quantify the preferences of each player over possible outcomes.

A central idea of game theory is that players must consider the decisions of other players. The best strategy for one player may therefore depend on the strategies they expect other players to choose.

Models of game theory tend to make assumptions about the players and their behaviour. For example, players may be assumed to act rationally and to choose strategies that they believe will lead to the most favourable outcome for themselves. These assumptions allow complex interactions to be represented mathematically.

In this project, we will introduce the basic mathematical concepts of game theory and applying them to The Prisoner's Dilemma. I will then explore how similar strategic reasoning can be applied to The Dark Forest Hypothesis, examining how assumptions about the behaviour and intentions of other players can influence the strategies considered optimal.

# 2. The Prisoner's Dilemma
In game theory, The Prisoner's Dilemma is a thought experiment that involves two players. These two players are thought of to be _rational_, meaning that one will always act in accordance with the aim to benefit oneself the most, or to achieve the least bad outcome. 

The "game" is as follows:

Two members of a criminal organisation are captured and imprisoned by the police. We will call them Prisoner A and Prisoner B. They are isolated individually, and have no means of communicating with one another. The police have insufficient evidence to convict them, and they plan to sentence the prisoners on a less harsh sentence where both will serve 1 year in prison. 

However, the prisoners are offered a choice - to stay silent (cooperate) or to testifies (defects). If Prisoner A testifies against Prisoner B, Prisoner A will go free, while the other is sentenced to 3 years in prison, and vice versa. However, if both prisoners testify against one another, they both will receive a sentence of 2 years. Needless to say, if they both stay silent, they both receive 1 year in prison. 

Both prisoners are told that they both were offered the same deal. They are unable to communicate with each other and have no way to know with absolute certainty what the other plans to choose. They are not told what the other has chosen until they have made their choice, at which point they cannot retract their decision. 

### Assumptions
- Each player is only concerned with maximising their own outcome.
- Both players understand the nature of the game
- Both players have no loyalty to each other
- There is no opportunity for reward or retribution outside of the game
- Both players aim to achieve the most desirable outcome for themselves. 

| Prisoner A \ Prisoner B | Prisoner B stays silent (cooperates) | Prisoner B testifies (defects) |
| --- | --- | --- |
| Prisoner A stays silent (cooperates) | Each serve 1 year | Prisoner B goes free, Prisoner A serves 3 years |
| Prisoner A testifies (defects) | Prisoner A goes free, Prisoner B serves 3 years | Each serve 2 years |

The optimal strategy:
From the point of view of Prisoner A, if B were to cooperate, ideally A should defect, so that he can go free rather than receiving a sentence of 1 year along with B. If B were to defect, Prisoner A should definitely defect as well, so that he serves 2 years rather than 3 years if he were to cooperate. In both cases, the better outcome is achieved by defecting. 

Therefore, regardless of what Prisoner B chooses, Prisoner A achieves a better individual outcome by defecting. The same reasoning can be applied from the point of view of Prisoner B. Thus, defecting is the _strongly dominant_ strategy for both players.

### Nash Equilibrium
A Nash equilibrium is a situation in which no player can improve their outcome by unilaterally changing their strategy, assuming the other players keep their strategies unchanged.

In The Prisoner's Dilemma, mutual defection forms a Nash equilibrium. Once both prisoners have chosen to defect, neither can improve their individual outcome by switching to cooperation: doing so would increase their sentence from two years to three years.

This creates a conflict between individual and collective interests. Mutual cooperation results in the lowest total sentence, with both prisoners serving one year, whereas mutual defection results in both prisoners serving two years. However, because defecting is the dominant strategy for each individual, rationally choosing to defect leads both players to the Nash equilibrium of mutual defection. 

Although mutual cooperation produces the best collective outcome, even if a player chooses to cooperate, they become vulnerable to the other player's decision. If the other player defects, the cooperating player ends up in a worse situation than they would have received by defecting themselves. This idea becomes important later when the Dark Forest is being discussed.



### Changing the payoffs
We will denote the outcomes of the game.\
$T = $ Temptation to defect \
$R = $ Reward for mutual cooperation\
$P = $ Punishment for mutual defection\
$S = $ Sucker's payoff, the result cooperating while the other defects

Here, the payoff is a reduction in jail sentence. \
Payoff is the negative of the prison sentence, the higher payoff the more desirable.

In the case of the class setup of The Prisoner's Dilemma, $T = 0$, $R = -1$, $P= -2$, $S = -3$.

In the classic Prisoner's Dilemma, the hierachy of payoffs is:
$$T > R > P > S$$

Regardless of what the other player chooses, defecting gives a greater payoff:
$$T > R$$
if the other cooperates, and
$$P > S$$
if the other defects.

Thus, the decision to defect is based on the ordering of the payoffs.


If we change the payoffs slightly such that

| Prisoner A \ Prisoner B | Prisoner B stays silent (cooperates) | Prisoner B testifies (defects) |
| --- | --- | --- |
| Prisoner A stays silent (cooperates) | Each serve 1 year | Prisoner B goes free, Prisoner A serves 3 years |
| Prisoner A testifies (defects) | Prisoner A goes free, Prisoner B serves 3 years | Each serve 3 years |

$T = 0$, $R = -1$, $P= -3$, $S = -3$.

Now, the punishment of mutual defection is equal to the sucker's payoff. 
$$T > R > P = S$$
If Prisoner B chooses to cooperate, Prisoner A should defect, going free rather than serving a year along with Prisoner B. If Prisoner B chooses to defect, Prisoner A's choice doesn't matter, Prisoner A will serve 3 years regardless of what he chooses. Defection is still preferred, though it is a _weakly dominant_ strategy. 

Here, we can call a strategy _strongly dominant_ if it gives a higher payoff in every situation, such as in the classic setup.\
We can call a strategy _weakly dominant_ if it gives at least an equal payoff in every situation, and at least one higher payoff in one.
In this modified Prisoner's Dilemma, defecting gives a higher payoff if the other cooperates, while giving the same payoff as cooperating if the other defects. 


We will consider another modified Prisoner's Dilemma:
| Prisoner A \ Prisoner B | Prisoner B stays silent (cooperates) | Prisoner B testifies (defects) |
| --- | --- | --- |
| Prisoner A stays silent (cooperates) | Each serve 1 year | Prisoner B serves 1 year Prisoner A serves 3 years |
| Prisoner A testifies (defects) | Prisoner A serves 1 year, Prisoner B serves 3 years | Each serve 3 years |

$T = -1$, $R = -1$, $P= -3$, $S = -3$.
$$T = R > P = S$$

Here, if Prisoner A defects, they serve 1 year if Prisoner B cooperates, and serve 3 years if Prisoner B defects. If Prisoner A cooperates, they serve 1 year if Prisoner B cooperates, and serve 3 years if Prisoner B defects. \
In this case, the payoff one receives is independent of the choice they make. It depends only on the choice the other player makes. Neither prisoner can improve their own outcome by changing their choice. If both prisoners cooperate, they can both obtain the shorter sentence of one year. However, cooperating does not guarantee that the other will cooperate as well, because neither cooperation nor defection provides an individual advantage. The payoff hierachy does not guarantee that mutual cooperation will occur.

In this modified Prisoner's Dilemma, my own outcome is entirely determined by the other player's decision. Since I do not gain any benefit from defecting, I might as well cooperate, giving the other player the opportunity to receive the shorter sentence as well. I can therefore hope that we reach mutual cooperation, and that the other player comes to the same conclusion.

Lastly, if it were to be the case that in a modified version of The Prisoner's Dilemma,
$$R > T > S > P$$
Mutual cooperation is the _strongly dominant_ strategy, because the reward for cooperating is greater than the temptation to defect, and also the sucker's payoff is also greater than the punishment for mutual defection.


### Repeated Games
If only a single round of The Prisoner's Dilemma is played, each player only has to consider the decision and outcomes made in that round. However, when considering repeated games, a player's current decision can affect how the other player plays in later rounds. 

Here, we will use the number of rounds, $N$, to be 10. $N=10$.

We will also define several strategies:

*Always Cooperate(AC)* : Cooperates every round.\
*Always Defects (AD)* : Defects every round.\
*Tit-for-Tat (TFT)* : Cooperates first, then repeats the choice the other player made in the previous round.\
*Grim Trigger (GT)* : Cooperates until the opponent defects once, and then GT will defect forever.

````python
import numpy as np
````

````python
def AD(opponent_history):
    return "D"

def AC(opponent_history):
    return "C"

def TFT(opponent_history):
    if not opponent_history:
        return "C"
    return opponent_history[-1]

def GT(opponent_history):
    if "D" in opponent_history:
        return "D"
    return "C"

def play_game(strategy_A, strategy_B, rounds=10):
    history = []

    history_A = []
    history_B = []

    for _ in range(rounds):
        move_A = strategy_A(history_B)
        move_B = strategy_B(history_A)

        history.append((move_A, move_B))

        history_A.append(move_A)
        history_B.append(move_B)

    return history
````

````python
payoffs = {
    ("C", "C"): (-1, -1),
    ("C", "D"): (-3, 0),
    ("D", "C"): (0, -3),
    ("D", "D"): (-2, -2)
}

def calculate_payoff(history):
    payoff_A = 0
    payoff_B = 0

    for moves in history:
        payoff = payoffs[moves]

        payoff_A += payoff[0]
        payoff_B += payoff[1]

    return payoff_A, payoff_B

def average_payoff(history):
    payoff_A, payoff_B = calculate_payoff(history)
    rounds = len(history)

    return payoff_A / rounds, payoff_B / rounds

````

````python
def g(p1,p2):
    return play_game(p1,p2,10)

strategies = [AD, AC, TFT, GT]

results = []

for p1 in strategies:
    row = []
    
    for p2 in strategies:
        payoff = average_payoff(g(p1, p2))
        row.append(payoff)
    
    results.append(row)

import pandas as pd
names = ["AD", "AC", "TFT", "GT"]
results_df = pd.DataFrame(
    results,
    index=names,
    columns=names
)

results_df
````


It can be observed that the dominant strategy earlier to always defect still carries over its property that is guarantees that it will result in a payoff either equal to or better than the opponents. If Player A chooses strategy AD, it guarantees that Player A gets at least $P$, while potentially getting $T$.

The results of the repeated games show that Tit-for-Tat (TFT) and Grim Trigger (GT) generally performed the best among the strategies considered. This allows both players to repeatedly receive the reward for mutual cooperation, $R$, rather than the lower payoff from mutual defection, $P$.

However, this does not mean that TFT or GT can be considered an optimal strategy. Their performance depends on the strategy of the opponent and on the payoff structure of the game. TFT and GT have an advantage in which cooperation is initially established, along with the possility for relataliation if they face defection. TFT reciprocates the opponent's previous action, while GT continues cooperating until the opponent defects, after which it defects permanently.

### Error
Previously, it was assumed that both players always correctly made the choice they intended, and that they knew the opponent's previous move perfectly. In reality, there will always be mistakes or misunderstandings in making choices. For example, a player may accidentally think that their opponent defected when they cooperated in reality. Perhaps if each player was a group of people, there may be miscommmunication between them and the person locking in their choice had misunderstood the group's intentions. Mistaking cooperation for defection greatly affects the outcomes for strategies such as TFT or GT. Because of this, we will introduce a probability that the choice is flipped. 

````python
from numpy import random
````

````python
def error(move, p):
    if random.uniform(0,1)<p:
        if move =="C":
            return "D"
        else:
            return "C"
    return move
````

````python
def play_game_error(strategy_A, strategy_B, rounds, p):
    history = []

    history_A = []
    history_B = []

    for _ in range(rounds):
        move_A = error(strategy_A(history_B),p)
        move_B = error(strategy_B(history_A),p)

        history.append((move_A, move_B))

        history_A.append(move_A)
        history_B.append(move_B)

    return history
````

````python
def comparison(strategy_A, strategy_B, rounds, p):
    print(strategy_A.__name__, "vs", strategy_B.__name__)
    print("No Error", play_game_error(strategy_A, strategy_B, rounds,0))
    print("p = ",p,play_game_error(strategy_A, strategy_B, rounds,p))
````

````python
comparison(TFT,TFT,10,0.05)
````

````python
comparison(TFT, GT, 10, 0.10)
````

````python
comparison(GT, GT, 10, 0.20)
````

In **TFT vs TFT**, the moment an accidental defection occurs, it sets off a chain of alternating defections. It is possible that if the error were to cause the defection in a round to flip back to cooperate, the chain could cease. Additionally, if the error causes a cooperation to flip to a defection while the opponent defects, meaning that both players defected in that round, this means the rest of the rounds will be mutual defections. However, mutual cooperation can be restored if the defection is fipped back to a cooperation, at which point mutual cooperation will resume until another error is made.

In **TFT vs GT**, the moment an error causes TFT to defect, it creates an endless chain of mutual defection. Even if the error flips one of the defections to a cooperation, either GT or TFT will resume defecting in the next round. GT will still continue to defect because TFT has defected previously, while TFT will continue to defect because GT will continue to defect. Even if the error causes both cooperations to flip to defections, GT will revert back to defecting in the next round, subsequently causing TFT to defect, resuming the mutual defection loop.

In **GT vs GT**, if an error causes a player to accidentally defect, it will cause a loop of mutual defections that cannot be broken even if a player accidentally cooperates.


````python
def monte_carlo(strategy_A, strategy_B, simulations=10000, rounds=100, p=0.05):

    total_A = 0
    total_B = 0

    for _ in range(simulations):

        history = play_game_error(strategy_A, strategy_B, rounds, p)

        payoff_A, payoff_B = calculate_payoff(history)

        total_A += payoff_A
        total_B += payoff_B

    average_A = total_A / simulations
    average_B = total_B / simulations

    return average_A, average_B
````

````python
print(monte_carlo(TFT, TFT, simulations=10, rounds=100, p=0.05))
print(monte_carlo(TFT, TFT, simulations=100, rounds=100, p=0.05))
print(monte_carlo(TFT, TFT, simulations=1000, rounds=100, p=0.05))
print(monte_carlo(TFT, TFT, simulations=10000, rounds=100, p=0.05))
````

````python
for p in [0, 0.01, 0.05, 0.10, 0.20]:
    result = monte_carlo(TFT, TFT, simulations=10000, rounds=100, p=p)
    print("p=",p, result)
````

# 3. The Dark Forest
The Dark Forest Hypothesis is the idea that alien civilisation are not rare in the universe, but we are unable to find them because they are hiding and staying silent out of fear. The Dark Forest Hypothesis serves as an explanation to the Fermi Paradox. 

The Fermi Paradox states that the universe is filled with very many planets, a lot of which are capable of sustaining life. Life is observed to expand to fill the space suitable for it. Despite that, we have found no traces of alien civilisations. 

The Dark Forest Hypothesis was introduced in science fiction book _The Dark Forest_ by author Liu Cixin. The Dark Forest Hypothesis says that alien civilisations will view other alien civilisations as a threat and thus will destroy any that they encounter if they have the capability to. 

The Dark Forest operates under these primary assumptions:
1. Survival is the primary need of all civilisations.
2. Civilisations expand in the cosmos, but the total matter in the universe is constant, meaning that expansion could create competition between civilisations.
3. The intentions of other civilisations cannot be reliably determined. One cannot establish with certainty whether the other is benevolent or malicious, even through communication. 
4. Even if Civilisation B appears benevolent based on the information available to Civilisation A, A cannot guarantee that B's intentions have remained unchanged since that information was emitted. Furthermore, the time required for messages to travel between the two civilisations means that their decisions must be made using potentially outdated information about one another.
5. Being attacked could result in catastrophic destruction. If one attacks first, the other may be unable to defend itself. The consequences of waiting too long could therefore be much worse than the costs of attacking.
6. A pre-emptive attack substantially reduces the threat. If one assumes that it has a sufficiently high chance of destroying the other before the other can retaliate, and that the benefits of doing so outweigh the risks of failure, retaliation and attacking a potentially peaceful civilisation.

For simplicity, we classify civilisations into two categories: _benevolent_ civilisations, which do not initiate attacks against others, and _malicious_ civilisations, which are willing to initiate attacks. This classification is a simplification; a civilisation's behaviour may depend on its circumstances and the information available to it.

Suppose that both civilisations possess the technological means to destroy one another. This is similar to nuclear deterrence on Earth, where nations possess weapons capable of causing catastrophic destruction, while still being vulnerable to attack. Under this assumption, revealing one's existence to another civilisation may carry a significant risk, particularly if its intentions and future technological capabilities are unknown. This raises the question of whether remaining silent, establishing communication, or launching a pre-emptive attack would be the most rational strategy.

Consider the following scenario: Civilisation A discovers the existence of Civilisation B. Suppose Civilisation B is not aware of the existence of Civilisation A. Civilisation A is now faced with 3 choices.

#### **Initiate Communication**

Civilisation A could try to establish communication wth Civilisation B. However, this presents a risk being that initiating communication will alert Civilisation B to their existence, and possibly Civilisation B eventually locating Civilisation A. 

If Civilisation B is _malicious_, they may destroy Civilisation A, possibly wiping out their whole species if they are planet-bound like us humans. In this case, maybe trying to establish cooperation leaves A in a situation where they might be wiped out. 

If Civilisation B is _benevolent_, they may exchange friendly communication for a period of time. However, because of the cosmic speed limit, sending messages back and forth may take hundreds of years if the civilisations are hundreds of light-years apart. This complicates things, because perhaps in the years it takes for messages to be sent and received, the socio-political climate of B may experience an overhaul, and they become a warmongering race. Therefore, they still present a threat to A, even if they may not have the means to attack A. The same reasning can be applied onto A by B. Suppose B is technologically inferior to A. B may communicate with A in a friendly manner but fear destruction. However, both civilisations starting out on unequal footing does not guarantee that one won't catch up or overtake the other in the future. As an example, consider that humans first developed firearms in the 10th Century, and later the first atomic bomb was used in the 20th Century, and later the hydrogen bomb was invented, thousands of times stronger that the atomic bomb. At that scale, the increase in power does not matter as much. The threat of destruction is the same. In the span of 1000 years, our technological capability for destruction exploded. Human destructive capabilities advanced from early firearms to nuclear and thermonuclear weapons over several centuries. There is no reason to assume that the same cannot happen on a galactic scale. Maybe B develops the means to destroy a planet sometime after A initiates communication with them. Again, the same reasoning can be applied the other way around. Thus, friendly communication is not guaranteed to last forever across the vast distance of space. 

Here, we introduce a concept called chain of suspicion. Suppose A is benevolent and does not wish to harm B.A still faces several uncertainties:
- A does not know whether B is benevolent.
- Even if B is benevolent now, A cannot guarantee that B will remain so indefinitely.
- If B discovers A, B faces the same uncertainty about A.  B has to consider the possibility that A may worry that B will eventually become malicious 
- B might therefore consider attacking A pre-emptively, not necessarily out of hostility, but to protect its own survival, eliminating the threat before they materialise

In this case, communication is dangerous as well.

In both cases, communication offers the possibility of cooperation while exposing A to an extinction level of risk. 

#### **Remain Silent**
In the Prisoner's Dilemma, staying silent,cooperating is the cooperative action. In the Dark Forest, cooperation is trying to establish friendly communication with the other. Unlike the Prisoner's Dilemma, here, there is another option besides to defect or to cooperate. Staying silent in the Dark Forest is a seperate option that aims to avoid revealing information. 
It is not too far fetched to conclude that if Civilisation A can discover the existence of Civilisation B, someday, B may discover A. This restarts the chain of suspicion. If B is malicious, they will attack A. Even if B is benevolent, they do not know if A is benevolent or malicious. Attempting communication is risky, and they can come to the same conclusion we discussed above. B may preemptively strike A. 

Assuming that B eventually discovers A, In the time that B takes to discover A, A may develop defensive technology, conceal its location, or monitor B's activities. However, this is still risky because B may exceed A in technological capabilities. Silence may be the safest choice if the probability of discovery is low. 

#### **Strike First**
If communication and remaining silent are both too dangerous to the survival of A's species. They are left with only one option: to attack. A pre-emptive strike may reduce B's opportunity to prepare or retaliate, especially if A has the capability to launch an attack at relativistic speeds. 

However, a surprise attack might fail, provoke retaliation, or reveal A's location to other civilisations. An attack could also be unnecessary if B is peaceful and never discovers A. Therefore, attack is not automatically the only rational option. However, under certain assumptions, a civilisation might judge a pre-emptive attack to offer the best chance of survival.

### The Survival Strategy

If civilisations are scattered throughout the universe, and revealing one's location could expose one to an existential threat, remaining undetected may be perceived as a form of protection. Civilisations may therefore avoid transmitting signals or revealing their locations, while treating the discovery of another civilisation as a potentially dangerous event, and attacking them if they possess the capability to.

Under the Dark Forest Hypothesis, the universe resembles a forest in which every inhabitant must consider the possibility that another could pose a threat. Revealing one's existence may therefore be perceived as dangerous, even when the intention is peaceful. If many civilisations adopt this reasoning, widespread silence could arise without requiring life to be rare in the universe.

The hypothesis offers one possible explanation for the Fermi paradox: civilisations may exist but remain undetected because they are reluctant to reveal their locations. However, this explanation depends on the assumptions made about survival, uncertainty, technological development and the behaviour of other civilisations. Whether these assumptions necessarily lead to aggression, or whether peaceful alternatives could be preferable, requires further examination.

##  The Game Theoretic Argument
One of the strongest points of The Dark Forest Hypothesis is its game theoretic explanation. 

In traditional game theory, the expected payoff $E$ of an action is calculated by multiplying the probability of each outcome by its value (payoff), and adding them all together:
$$E = \sum (\text{Probability} \cdot \text{Payoff})$$


Let
$q = \text{probability that B discovers A if A remains silent}$

$r_A = \text{probability that A's attack on B succeeds, otherwise it failed}$

$r_B=\text{probability that B's attack on A succeeds, otherwise it failed, or was defended against}$

$s_A = \text{probability that A survives an attack by B}$

$s_B = \text{probability that B survives an attack by A}$

In the classic Dark Forest Hypothesis, the primary assumption is that $L\rightarrow \infty$. It is also implicitly assumed that if a malicious civilisation were to attack another civilisation, it would lead to the eradication of the attacked. ($r_A=r_B=1$, and $s_A=s_B=0$)
We assume that $r_A=r_B=1$, and $s_A=s_B=0$, meaning that any attack on any other civilisation is successful, and that suffering an attack leads to maximum loss. Also, $P(\text{B attacks}), q>0$, meaning that the possibility that B is malicious cannot be ruled out.
We define $D$ to be the expected cost for A suffers if A survives B's attack, and $V$ to be the value of cooperation.


##### **Communication**
$$
\begin{aligned}
E(\text{Communication}) &= \text{Probability of cooperation} \cdot \text{Value of cooperation} \\
&+ \text{Probability of attack} \cdot \text{Cost of being attacked} + \text{Probability of being ignored} \cdot 0\\
&= P(\text{B cooperates | B receives communication}) \cdot V + P(\text{B attacks | B receives communication})\cdot -\infty
\end{aligned}
$$
Under the assumption that $s=0$, the term $(1-s)\cdot -\infty$ dominates the equation. because $V$ is finite, $E(\text{Communication})$ collapses to $-\infty$

$$E(\text{Communication}) = (-\infty)$$


##### **Staying Silent**
$$E(\text{Staying Silent}) =  (1-q) \cdot 0 + q \cdot E(\text{Discovered})$$

$$
E(\text{Discovered}) =P(\text{B attacks | B discovers A}) \cdot (-\infty) + P(\text{B cooperates | B discovers A}) \cdot V
$$

Again,  the term $(1-s)\cdot -\infty$ dominates the equation.
$$E(\text{Discovered}) = (-\infty)$$
$$\implies E(\text{Staying Silent}) = (-\infty)$$


##### **Striking First**
$$
E(\text{Attack}) = -C
$$
where $C$ is the economical cost to attack.

Under the assumptions of the Dark Forest where a strike is guaranteed to succeed $(r_A=r_B=1)$, and it is impossible to survive a strike $(s_A =s_B=0)$, $L\rightarrow \infty$, calculating the expected values show that **striking first** has the highest payoff of the choices.

## Varying the Probabilities
It can be seen clearly in the expected value of attacking that The Dark Forest Hypothesis assumes that launching a preemptive strike is guaranteed to eliminate the threat. However, if that assumption isn't made, it may be possible that striking first isn't the only rational choice. 

The classic model relies on three extreme assumptions: 
$r=1$, $s=0$, and $L\rightarrow \infty$. This implicitly assumes that both civilisations have the means to destroy one another, and that an attack fully destroys the other, leaving them no opportunity for retaliation, leading to the incentive to strike first. These assumptions capture the bleakness of the Dark Forest, but they do not reflect the physics of asymmetric warfare. A very advanced civilization attacked by a weaker one does not face extinction. Earlier, parameters $s$,$r$ etc. were loosely defined. To build a more realistic model, we will redefine the offensive and defensive parameters.

$r =$ Attack success. 

The probability that a civilization's weapon system successfully bypasses the defender's defences and lands a strike.

$s= $ Survival probability.

The probability that a civilization avoids major loss ($L$) when struck, suffering only minor loss ($D$). It is dependent on the technological gap between the defender and the attacker.

$L= $ Major loss

The irreversible destruction of a civilization's capacity to function as an independent interstellar entity. The physical manifestation of L depends on the civilization's distribution. For a weak civilization, concentrated on a single planet, L is biological extinction, the death of the species. For an advanced civilisation distributed across a galaxy, L cannot mean extinction. L is the destruction of a planet or solar system. It is a costly loss, but the galactic species survives.

$D= $ Minor loss

The cost suffered when a civilization survives a strike. The destruction of a minor asset (e.g., a base, a space station, or a colony). It is painful but does not threaten the civilization's existence.

We simplify the model with these assumptions: 
1. B is will attack with probability $p_{attack}$, cooperate with probability $p_{coop}$ or stay silent with probability $p_{silent}$.
2. B attacks as soon as it is aware of A's existence if it is hostile. B cooperates if it receives communication from A, or if it discovers a silent A and chooses to be benevolent.
3. If A stays silent, and is discovered by B, and if B chooses to cooperate, A will automatically accept it leading to mutual cooperation, because silence is assumed to be out of fear of uncertainty of B's nature. B initiating cooperation therefore reveals their nature of benevolence. 
4. A failed attack provokes retaliation, regardless of whether B is malicious or benevolent. 

We set the general conditions $s \geq 0$ and $r \leq 1$. This can be considered a realistic scenario as it encompassing all possible combinations of offensive and defensive capability. A can survive an attack from B, but A's attack on B is not guaranteed to succeed. 

$$
\begin{aligned}
E(\text{Communication}) &= P(\text{B cooperates | B receives communication}) \cdot V \\
&+ P(\text{B attacks | B receives communication}) \cdot [r_B(-L(1-s_A) -s_AD)+(1-r_B)0] \\
&+ P(\text{B silent | B receives communication}) \cdot 0
\end{aligned}
$$
Let $E^A_B$ be expected value of a successful attack from B onto A. $E^A_B = (-L(1-s_A) -s_AD)$
$$\boxed{E(\text{Communication})=p_{coop}V + p_{attack} (r_B\cdot E^A_B)}$$


$$E(\text{Staying Silent}) =  (1-q) \cdot 0 + q \cdot E(\text{Discovered})$$

$$\begin{aligned}
E(\text{Discovered}) &=P(\text{B attacks | B discovers A}) \cdot [r_B \cdot E^A_B] \\
&+ P(\text{B cooperates | B discovers A}) \cdot V+ P(\text{B silent | B discovers A}) \cdot 0\\
\\
&= p_{attack}[r_B\cdot E^A_B] + p_{coop}V
\end{aligned}$$
$$\boxed{E(\text{Staying Silent}) = q[p_{coop}V+p_{attack}(r_B)(E^A_B)]}$$

$$
E(\text{Attack}) = -C + (1-r_A)[r_B(-L(1-s_A) -s_AD)]
$$
$$
\boxed{E(\text{Attack})=-C + (1-r_A)(r_B)(E^A_B)}
$$

**Communication vs Staying silent**

We want to find the condition such that
$$E(\text{Communication}) > E(\text{Staying Silent})$$
We define $\Delta_A = - E^A_B$ to be the cost for A when A is successfully attacked. $\Delta_A = L(1-s_A) +s_AD$

$$
\begin{aligned}
p_{coop}V + p_{attack}\cdot  r_B \cdot E^A_B&>q[p_{coop}V+p_{attack}\cdot r_B \cdot E^A_B]\\
(1-q)[p_{coop}V+p_{attack}\cdot r_B\cdot E^A_B]&>0\\
p_{coop}V &> p_{attack}\cdot r_B\cdot \Delta_A
\end{aligned}
$$
Thus,
$$\boxed{E(\text{Communication}) > E(\text{Staying Silent})\implies p_{coop}V > p_{attack}\cdot r_B\cdot \Delta_A}$$

**Communication vs Attacking**
$$E(\text{Communication}) > E(\text{Attack})$$

$$
\begin{aligned}
p_{coop}V + p_{attack} \cdot r_B\cdot E^A_B&>-C + (1-r_A) \cdot r_B \cdot E^A_B\\
E^A_B\cdot r_B(p_{attack} - (1-r_A))&>-C-p_{coop}V\\
\Delta_A \cdot r_B(p_{attack} - (1-r_A))&<C+p_{coop}V
\end{aligned}
$$
$$\boxed{E(\text{Communication}) > E(\text{Attack})\implies\Delta_A \cdot r_B[p_{attack} - (1-r_A)]<C+p_{coop}V}$$
The difference between expected cost if attacked while communicating and the expected cost of retaliation has to be less than the sum of cost of attacking and expected value of cooperation. 

If  $p_{attack} - (1-r_A)<0$, that is, $p_{attack} <1-r_A$, the probability that A will be attacked by B is less than the probability that A will miss their attack, then **communication is always better than attacking** because the left hand side of the expression is negative.

If $p_{attack} - (1-r_A)>0$, that is, $p_{attack} >1-r_A$, the probability that A will be attacked by B is greater than the probability that A will miss their attack, then communication is only better than attacking if $\Delta_A \cdot r_B(p_{attack} - (1-r_A))<C+p_{coop}V$.

**Attacking vs Staying Silent**
$$
\begin{aligned}
E(\text{Attack})&>E(\text{Staying Silent})\\
-C + (1-r_A) \cdot r_B \cdot E^A_B&>q(p_{coop}V+p_{attack}\cdot r_B\cdot E^A_B)\\
E^A_B\cdot r_B(q\cdot p_{attack} - (1-r_A))&<-C-q\cdot p_{coop}V\\
\Delta_A \cdot r_B(q\cdot p_{attack} - (1-r_A))&>C+q\cdot p_{coop}V
\end{aligned}
$$
Hence,
$$\boxed{E(\text{Attack})>E(\text{Staying Silent})\implies \Delta_A \cdot r_B(q\cdot p_{attack} - (1-r_A))>C+q\cdot p_{coop}V}$$

The difference between expected cost if attacked after being discovered while staying silent and the expected cost of retaliation has to be less than the sum of cost of attacking and expected value of cooperation from B if A stays silent. 

If  $q\cdot p_{attack} - (1-r_A)<0$, that is, $q\cdot p_{attack} <1-r_A$, the probability that A will be discovered and attacked by B is less than the probability that A will miss their attack, then **attacking will never be better than staying silent** because the left hand side of the expression is negative, while the right hand side is positive.

We will observe the conditions for attacking to be the better strategy.
$$E(\text{Attack})>E(\text{Communication})  \implies \Delta_A \cdot r_B[p_{attack} - (1-r_A)]<C+p_{coop}V$$
Thus, $$\Delta_A >\frac{C+p_{coop}V}{r_B(p_{attack} - (1-r_A))}$$ ( $p_{attack} - (1-r_A)>0$, otherwise communication is always better than attacking ).

Also,
$$E(\text{Attack})>E(\text{Staying Silent})  \implies \Delta_A \cdot r_B[p_{attack} - (1-r_A)]<C+p_{coop}V$$
Thus, $$\Delta_A >\frac{C+q\cdot p_{coop}V}{r_B(q\cdot p_{attack} - (1-r_A))}$$ ( $q\cdot p_{attack} - (1-r_A)>0$, otherwise staying silent is always better than attacking ).

**Claim:** 
$$\frac{C+q\cdot p_{coop}V}{r_B(q\cdot p_{attack} - (1-r_A))}>\frac{C+p_{coop}V}{r_B(p_{attack} - (1-r_A))}$$
**Proof:**
$$
\begin{aligned}
(C+q\cdot p_{coop}V)(p_{attack} - (1-r_A))&>(C+p_{coop}V)(q\cdot p_{attack} - (1-r_A))\\
C\cdot p_{attack} -C(1-r_A) +q\cdot p_{coop}V\cdot p_{attack} -q\cdot p_{coop}V\cdot (1-r_A)&>C\cdot q \cdot p_{attack} -C(1-r_A) + p_{coop}V\cdot q\cdot p_{attack} - p_{coop}V\cdot (1-r_A)\\
(1-q)C\cdot p_{attack}+(1-q)p_{coop}V\cdot (1-r_A) >0\\
C\cdot p_{attack}+p_{coop}V\cdot (1-r_A)>0
\end{aligned}
$$
This is always true. Therefore,
$$\boxed{E(\text{Attack})>E(\text{Staying Silent})  \implies E(\text{Attack})>E(\text{Communication})}$$

#### Comparing the conditions

| Comparison | Condition |
|---|---|
|$E(\text{Communication}) > E(\text{Staying Silent})$| $p_{coop}V > p_{attack}\cdot r_B\cdot \Delta_A$|
|$E(\text{Communication}) > E(\text{Attack})$|$\Delta_A \cdot r_B[p_{attack} - (1-r_A)]<C+p_{coop}V$|
|$E(\text{Attack})>E(\text{Staying Silent})$|$\Delta_A \cdot r_B(q\cdot p_{attack} - (1-r_A))>C+q\cdot p_{coop}V$|

## The Conditions for Darkness
The Dark Forest is a state in which communication is never the dominant strategy. In this state, a civilization's only rational choices are to Hide or to Attack. The table above shows the mathematical conditions for this state, which fall into two distinct regimes:
1. The predators: $q \cdot p_{attack}>1−r_A$ and $\Delta_A >\frac{C+q\cdot p_{coop}V}{r_B(q\cdot p_{attack} - (1-r_A))}$ (Attacking is dominant).
2. The prey: $p_{coop}V < p_{attack}\cdot r_B\cdot \Delta_A$ and $q\cdot p_{attack} - (1-r_A)<0$ (Silence is dominant).


Where $\Delta_A = L(1-s_A) +s_AD$.
When these conditions are met, the universe becomes a dark forest. However, the strategy that keeps the forest dark depends entirely on the civilization's offensive and defensive capability. 

The classic hypothesis assumes that $L\rightarrow \infty$, $s=0$ and $r=1$. We have derived the general conditions for the Dark Forest Hypothesis as shown in the table above. In the next section, we will apply our variables onto the Kardashev scale to understand how different types of civilizations behave.

## The Kardashev Scale
The Kardashev scale classifies civilizations by their technological capability and energy consumption. The concept of this scale as well as the physical meaning of parameters discussed in this section are all speculative but useful to hypothesize the way in which The Dark Forest manifests in technological superior races. 

We will analyse the choices Civilisation A has depending on how technologically advanced they could hypothetically be.

### **Type I Civilisations, $K_A=1$ (Planetary):**
Generally speaking, _Type I civilisations_ are classified by their ability to access all the energy available on its planet and store it for consumption. 

If A were a _Type I civilisation_, they would be rather close but ahead of humanity in terms of technological capability (though we are around 0.7 on the scale) with limited energy harnessing capabilities. They are bound to one planet and have limited capability to withstand an interstellar strike from a potentially more advanced alien civilisation. 

For a Type I civilisation, 

- $s_A \approx 0$ against superior civilisations.

- $\Delta_A \rightarrow \infty$, 
- $r_A \approx 0$ It is difficult to attack across the vast cosmos .
- $C$ very large.
- $L \rightarrow \infty$. Total extinction.
- $D$ very large

Since $r_A\approx 0$, $(1-r_A) \approx 1$. 
They find that $q\cdot p_{attack} -(1-r_A) <0 \implies q\cdot p_{attack}<1$ is almost always true.

Since $\Delta_A \rightarrow \infty$, $p_{coop}V<p_{attack}\cdot r_B \cdot \Delta_A$ is also true.

This satisfies the conditions for the prey regime. 

Thus, **The Dark Forest holds true for weak Type I civilisations**, manifesting in the form of **silence**.

### **Type II Civilisation, $K_A = 2$ (Stellar):** 
_Type II civilisations_ can directly consume a star's energy.

If A were a Type II civilisation, we can model them as having:

- $0<r_A<1$. They may be able to reliably launch a strike targetted at a location far away in space.
- $0<s_A<1$. They may have some defensive method.
- $C$ moderate
- $L$ very large. Could be thought of as losing a home planet or core star system.
- $D$ large. The loss of a major colony.

Type IIs have overall moderate offensive and defensive capabilities. Despite $L$ and $D$ being large, $V$ also has the potential to be quite large if they manage to cooperate. 

The Type II civilization are the balance point in the universe. None of the 3 strategies are obviously dominant for them. Their decision can be justified depending on their values of $r_A$, $s_A$, and $\Delta_A$.



### **Type III Civilisations, $K_A =3$ (Galactic):** 
A Type III civilisation has the capability to harness all energy in their galaxy from any source, from stars to black holes. 

A Type III civilisation has the technological means to attack and defend themselves. If A were a Type III civilisation, they face very low risk when an attack on them is successful, because they have spread out and inhabit multiple locations in the galaxy. The loss of one of their bases to a strike isn't as devastating. They have:
- $s_A \approx 1$ against inferior civilisations
- $r_A \approx 1$ their weapons are guaranteed to hit any target
- $C \rightarrow 0$. For a galactic civilisation, the cost to develop an interstellar weapon is trivial.
- $L$ moderate to large. Could be the loss of a major planetary system, or a major space city. It is only a fraction of the numerous networks spread out across the galaxy.
- $D$ low. The repair cost is negligible. 

**Finding an inferior civilisation**

Condition 1 is easily met because $(1-r_A) \approx 0$, and $\Delta_A = L(1-s_A) +s_AD \implies \Delta_A \approx D$. 

$\Delta_A >\frac{C+q\cdot p_{coop}V}{r_B(q\cdot p_{attack} - (1-r_A))}$ is easily met because $C \rightarrow 0$ and $V\approx 0$ even if they cooperate with a weaker civilisation. 

For an advanced Type III civilisation, if the civilisation they discovered is weaker than them, both $D \rightarrow 0 \implies \Delta_A \approx E^A_B \approx 0$ and $V \rightarrow 0$. If Civilisation A is a Type III civilisation, and Civilisation B is in its infant stage, neither the risk of attacking nor the benefit of cooperation matter to A.
$$E(\text{Communication})=p_{coop}V + p_{attack} (r_B)(E^A_B) \approx 0$$
$$E(\text{Staying Silent}) = q[p_{coop}V+p_{attack}(r_B)(E^A_B)] \approx 0$$
$$E(\text{Attack})=-C + (1-r_A)(r_B)(E^A_B) \approx 0$$

Civilisation A is indifferent to B. Neither choice has significant value to A, because B neither poses a threat nor does it offer value in cooperation. Therefore, A may simply strike out of convenience because $C \rightarrow 0$. 

Additionally, if left unchecked, B could grow technologically and pose a threat to A in the future $D \uparrow$, incentivising A to pre-emptively strike. This idea will be further discussed in a later section.

**Finding an equal**

If Civilizations A and B are roughly technologically equal (both Type III), their capabilities are very similar. Their probability of surviving each other's strikes is roughly $s_A\approx s_B \approx 0.5$. B poses a profound threat to A not because they can drive A to biological extinction, but because they possess weapons capable of inflicting massive material damage. A faces a 50% chance of suffering a major loss $L$ (the destruction of a planetary system) and a 50% chance of suffering minor damage $D$. 

On the other hand, the potential value of sharing technology and resources, $V$ is rather large. However, it is not difficult to imagine that $\Delta_A$ and $V$ scale disproportionately. Because both civilizations have already conquered their respective galaxies, the marginal benefit of cooperation, $V$, leads to diminishing returns. There is not much a peer can offer that A cannot eventually obtain on their own. 

Because $r_A \rightarrow 1$,$r_B \rightarrow 1$ and $C \rightarrow 0$, 
$$
\begin{aligned}
E(\text{Attack})>E(\text{Staying Silent})&\implies \Delta_A >\frac{C+q\cdot p_{coop}V}{r_B(q\cdot p_{attack} - (1-r_A))}\\
&\implies \Delta_A > \frac{q\cdot p_{coop} V}{q\cdot p_{attack}}\\
&\implies \Delta_A > \frac{p_{coop} V}{p_{attack}} \\
&\implies \boxed{p_{attack} \Delta_A > p_{coop} V}
\end{aligned}
$$

The expected loss of being attacked, $p_{attack}\Delta_A$ remains catastrophic compared to $p_{coop}V$, which shows that attacking is preferred over staying silent. Additionally, as we have proved earlier, this also implies that attacking is also better than communicating. 

Even if A chooses to stay silent, A can deduce that interstellar war is inevitable. B will eventually discover A, and upon doing so, will face the exact same dilemma. This is the chain of suspicion: A knows that B is reasoning exactly as A is. Therefore, A may choose to strike preemptively. By starting the war early while still undetected, A gains a decisive first-striker advantage by forcing B into a reactive position and damaging B's resources while they are still undetected. 

**The Dark Forest holds true for Type III civilisations**, who serve as apex predators hunting on weaker civilisations. 

The Dark Forest is a universal state. The universe is silent because the weak are hiding, the strong are effortlessly destroying the weak, and the apex predators are preparing to strike first upon finding another. 

## The Berserker 

There is a possibility there exists a civilization that attacks unconditionally, regardless of the target's actions. We will call this a _berserker_. It does not communicate, does not stay silent, and does not respond to signals. In our model, this is the case where $p_{attack}=1$.

The general idea is that it is pointless to attempt to communicate with a berserker. Communication reveals your location without any possibility of changing B's behaviour. A berserker attacks regardless. So the act of communicating only ensures that A is attacked, whereas staying silent gives A a chance ($1-q$) of remaining undiscovered.

$$E(\text{Communication})=p_{coop}V + p_{attack} (r_B\cdot E^A_B) = r_B\cdot E^A_B = -r_B\cdot \Delta_A$$
$$E(\text{Staying Silent}) = q[p_{coop}V+p_{attack}(r_B)(E^A_B)] = q \cdot r_B \cdot E^A_B=-q \cdot r_B \cdot \Delta_A$$

$$\implies qE(\text{Communication})=E(\text{Staying Silent})$$

Since $q<1$ and $E(\text{Communication}),E(\text{Staying Silent})<0$
$$E(\text{Communication}) < E(\text{Staying Silent})$$

If A counters a berserker type civilisation, staying silent is always better than communicating, regardless of any other parameters. 

Thus, A has to choose between attack and silence. As shown above, Type I civilisations is forced into silence, and Type III civilisations attack, while Type II civilisations have to choose depending on their attack and survival capabilities.

## Preemptive Strikes
Our model assumes a civilisation's type is fixed. Liu Cixin's science fiction novel _Death's End_ provides a concrete illustration of pre-emptive strike logic. In the story, the character Singer destroys the solar system not because humanity is currently a threat, but because it might become one. Across the cosmos, communication only takes the form of sending messages back and forth with potentially a few hundred years while the messages travel through the void of space.

Consider the following situation: A deduces that they are currently technologically superior to B in both offence and defense (A is a _Type III civilisation_, B is a _Type I civilisation_). This is represented by $r_A \approx 1$, $s_B \approx 0$ (A can destroy B now) and $r_B \approx 0$,$s_A \approx 1$ (B cannot destroy A now). 

However, the threat is not in the present, but in the future. We will assume that with probability $g$, B grows to become a Type III civilization, catching up to A in terms of technological development, and that it detects A in the future with probability $q$. If B finds A, it may be attack (probability $p_{attack}$), initiate communication (probability $p_{coop}$), or ignore A (probability $p_{silent}$). If B does not grow, it remains weak and irrelevant.

$g=$ Probability that B grows into a Type III civilization (or a competitor).

$q=$ Probability that B discovers A in the future if A stays silent and ignores B.

$p_{attack}=$ Probability that B is will attack in the future.

$p_{coop}=$ Probability that B will cooperate in the future.

$p_{silent=}$ Probability that B ignores A if they discover A in the future.

$C=$ Cost to A of striking B pre-emptively right now.

$W=$ Cost of a future interstellar war if B discovers A and is hostile.

$V=$ Value of cooperation if B discovers A and is benevolent.

Note: Earlier, $D$ represents the immediate damage from being struck, plus the cost of rebuilding. It does not include the cost of an interstellar war. In this section, we introduce $W$ as the expected cost of a future war that occurs after B has grown and discovered A. This seperates the immediate costs of an attack from the longer-term costs of war.

A is presented with two choices: strike now or stay silent.
$$E(\text{Attack}) = -C \approx 0 $$
$$E(\text{Staying silent}) =(1-g)0 + gq(-p_{attack}W + p_{coop}V) =gq(-p_{attack}W + p_{coop}V)$$

Note: Communication is always worse than staying silent because A gains nothing from cooperating with a weaker civilisation $V \approx 0$, while the cost of a future war, $W$, could be quite large. Communicating effectively makes it so $q=1$. Thus, $E(\text{Staying silent}) > E(\text{Communication})$.


By considering the necessary conditions such that 
$$E(\text{Attack})>E(\text{Staying silent})$$
We have that
$$0>gq(-p_{attack}W + p_{coop}V) \approx gq(-p_{attack}W)$$


It is not unreasonable to assume that growth is guaranteed. Any civilization that survives will eventually expand and become a Type III civilization. We can assume that $g>0$ and $q>0$, because the threshold simplifies to 
$$
\begin{aligned}
0&>-p_{attack}W\\
p_{attack}W&>0\\
\end{aligned}
$$
This is always true. This shows that the cost of a future war with a peer far outweighs the cost of destroying a weaker civilisation.

This analysis reveals that the Dark Forest Hypothesis does not require infinite war costs or universal hostility. It requires only a specific asymmetry: the cost of striking now, $C$ must be small relative to the expected cost of a future war, $W$.


## The Reality of The Dark Forest
We have analyzed the Dark Forest Hypothesis across the entire spectrum of civilizational capability. Our analysis confirms that the hypothesis holds universally: for every type of civilization, communication is never the optimal strategy, leaving only silence or attack. However, the specific manifestation of this universal silence differs fundamentally across the Kardashev scale.

Ultimately, the Dark Forest is not dark because every civilization is a berserker. It is dark because the mere possibility of being discovered by a hostile civilization carries consequences from immediate annihilation ($L$) or heavy damage ($D$), to a future war ($W$). That almost always exceed the expected value of cooperation ($V$). The forest is dark because the risk of being wrong is simply too high to bear.

## Modelling Galactic Asymmetric Warfare
To quantify a civilization's technological and military capability, we define $R$ to represent the civilisation's resources. $R$ measures a civilization's total accessible energy and material resources. We also define a map from $R$ to the Kardashev scale $K$, in the form $K = \log_{1000}R$. Thus,

|Kardashev Type|$R$|
|---|---|
|$0$|$1$|
|$1$|$10^3$|
|$2$|$10^6$|
|$3$|$10^9$|

$R$ is such that
- $r$, $s$, $C$, $L$, $D$, and $V$ are all dependent on $R$.

- When a civilisation is attacked, R decreases. They need to use their resources to rebuild.

- When a civilisation achieves mutual cooperation, R increases. They gain energy and resources.

- If $R \leq 0$, the civilisation is treated as extinct.

We will model the parameters as such:
- $K_A = \log_{1000}(R_A)$ is the Kardashev Type.
- $r_A = \min(0.001\cdot 10^{K_A},1)$ is the probability A's strike hits. Scales with A's technological capability.
- $s_A = \frac{1}{1 + 10^{(K_B - K_A)}}$ is the probability A survives B's strike, depending on the technological gap.
- $L = \frac{R_B}{10^{K_A}}$ is the cost of the destruction of a planet or system. It scales with the attacker's wealth, and reduced by the civilisation's distribution
- $D = \min(\frac{R_B}{(K_A+1) \cdot 10^{K_A}},0.99R_A)$ is the cost of the destruction of a minor base
- $V = \sqrt{R_A \cdot R_B}$ is the value of mutual cooperation
- $C_A = \frac{0.01R_A}{10^{K_A-1}}$ the economic cost to build and use a weapon
- $q = \min(0.01(1+2K_A)(1+2K_B),1)$ the probability that B will discover A if A stays silent

**Growth Factor**

Each round can be thought of as taking place over the span of a few hundred years if the two civilisations are a few hundred light years apart. Thus, we introduce a growth factor each round such that $R_{new} = R_{old} \cdot (1+g)$
We model $g$ as
$$g= \frac{9}{10^K}$$
so that infant civilisations grow quickly as they begin to tap into the resources available to them, whereas growth is much lower for more advanced civilisation who have accessed all resources available to them. 

This will make it so that if a Type 0 or Type I is allowed to grow freely, they will easily climb the resource ladder in a few rounds. This is technological explosion. 

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. CONSTANTS AND PARAMETERS

NUM_CIVS = 1000
NUM_ROUNDS = 100
NUM_TRIALS = 50      # Number of Monte Carlo runs
SPAWN_RATE = 0.5    # 50% chance to spawn a new Type 0 civ per round

INITIAL_STRATS = ['Communicate', 'Silent', 'Attack']

P_ATTACK = 1/3
P_SILENT = 1/3
P_COOP = 1/3

# 2. PARAMETER FUNCTIONS
def K(R):
    return np.log(max(R, 1e-3)) / np.log(1000)

def r(R_A):
    return min(0.001 * (10**K(R_A)), 1.0)

def s(R_A, R_B):
    return 1 / (1 + (10**(K(R_B) - K(R_A))))

def L(R_A, R_B):
    return R_B / max(K(R_A), 1.0)

def D(R_A, R_B):
    raw_damage = R_B / ((K(R_A) + 1) * max(K(R_A), 1.0))
    return min(raw_damage, 0.99 * R_A)

def C(R_A):
    return (0.01 * R_A) / (10**(K(R_A) - 1))

def V(R_A, R_B):
    return 0.01 * np.sqrt(R_A * R_B)

def q(R_A, R_B):
    return min(0.01 * (1 + 2*K(R_A)) * (1 + 2*K(R_B)), 1.0)

# 3. INTERACTION PAYOFFS

def resolve_interaction(strat_A, R_A, strat_B, R_B, A_detects_B, B_detects_A):
    r_A = r(R_A); s_A = s(R_A, R_B); C_A = C(R_A)
    r_B = r(R_B); s_B = s(R_B, R_A); C_B = C(R_B)
    
    v_val = V(R_A, R_B)
    L_from_B = L(R_A, R_B); D_from_B = D(R_A, R_B) 
    L_from_A = L(R_B, R_A); D_from_A = D(R_B, R_A)

    payoff_A = 0.0
    payoff_B = 0.0

    # --- 1. A's Action ---
    if A_detects_B:
        if strat_A == 'Communicate':
            # A BROADCASTS. B AUTOMATICALLY RECEIVES THE MESSAGE.
            # B now MUST choose an action based on its strategy.
            if strat_B == 'Attack':
                # B is malicious! B attacks the transmitter.
                if np.random.uniform(0, 1) < r_B: 
                    if np.random.uniform(0, 1) < s_A: payoff_A = -D_from_B
                    else: payoff_A = -L_from_B
                    payoff_B = -C_B 
            elif strat_B == 'Silent':
                # B ignores the message.
                payoff_A = 0; payoff_B = 0
            elif strat_B == 'Communicate':
                # B responds with cooperation.
                payoff_A = v_val; payoff_B = v_val
                
        elif strat_A == 'Silent':
            # A stays silent. B will act in block 2 only if B detects A.
            pass 
            
        elif strat_A == 'Attack':
            # A attacks B.
            if np.random.uniform(0, 1) < r_A: 
                payoff_A = -C_A
                if np.random.uniform(0, 1) < s_B: payoff_B = -D_from_A
                else: payoff_B = -L_from_A
            else: 
                # A misses. B retaliates.
                if np.random.uniform(0, 1) < r_B: 
                    if np.random.uniform(0, 1) < s_A: payoff_A = -D_from_B
                    else: payoff_A = -L_from_B
                    payoff_B = -C_B

    # --- 2. B's Action (If A was Silent or missed) ---
    if B_detects_A:
        if strat_B == 'Attack' and strat_A != 'Attack':
            # B attacks A (because A was silent).
            if np.random.uniform(0, 1) < r_B: 
                if np.random.uniform(0, 1) < s_A: payoff_A = -D_from_B
                else: payoff_A = -L_from_B
                payoff_B = -C_B
                
        elif strat_B == 'Communicate' and strat_A == 'Silent':
            # B communicates with a Silent A.
            payoff_A = v_val; payoff_B = v_val

    return payoff_A, payoff_B

# 4. SIMULATING
def run_simulation():
    population = []
    for _ in range(NUM_CIVS):
        strat = np.random.choice(INITIAL_STRATS)
        wealth = np.random.choice([1, 1000, 1e6, 1e9])
        population.append([strat, wealth])

    history = {
        'Communicate': [], 'Silent': [], 'Attack': [],
        'Type 0': [], 'Type I': [], 'Type II': [], 'Type III': [],
        'T2_Comm': [], 'T2_Sil': [], 'T2_Att': [],
        'T3_Comm': [], 'T3_Sil': [], 'T3_Att': []
    }

    for round_num in range(NUM_ROUNDS):  # <--- FIXED: Indented inside function
        # 1. Pairing random civilisations
        indices = np.random.permutation(len(population))
        payoffs = np.zeros(len(population))
        
        for i in range(0, len(population) - 1, 2):
            idx_A = indices[i]
            idx_B = indices[i+1]
            strat_A, R_A = population[idx_A]
            strat_B, R_B = population[idx_B]
            
            q_val = q(R_A, R_B)
            A_detects_B = np.random.uniform(0, 1) < q_val
            B_detects_A = np.random.uniform(0, 1) < q_val
            
            if A_detects_B or B_detects_A:
                # Randomize who goes first if both detect each other
                if A_detects_B and B_detects_A:
                    if np.random.uniform(0, 1) < 0.5:
                        # A goes first
                        pay_A, pay_B = resolve_interaction(strat_A, R_A, strat_B, R_B, True, True)
                    else:
                        # B goes first! Swap A and B.
                        pay_B, pay_A = resolve_interaction(strat_B, R_B, strat_A, R_A, True, True)
                elif A_detects_B:
                    pay_A, pay_B = resolve_interaction(strat_A, R_A, strat_B, R_B, True, False)
                else: # B_detects_A must be true
                    pay_B, pay_A = resolve_interaction(strat_B, R_B, strat_A, R_A, True, False)
                
                payoffs[idx_A] += pay_A
                payoffs[idx_B] += pay_B

        # 2. Payoffs, Growth & Extinction
        new_population = []
        for i, civ in enumerate(population):
            strat, R_old = civ
            R_new = R_old + payoffs[i]
            
            if R_new > 0:
                K_val = K(R_new)
                g = 9 / (10 ** K_val)
                growth = R_new * g
                R_final = R_new + growth
                new_population.append([strat, R_final])

        # 3. Spawn new Type 0s with a random strategy
        if np.random.random() < SPAWN_RATE:
            new_strat = np.random.choice(INITIAL_STRATS)
            new_population.append([new_strat, 1.0])

        population = new_population
        
        # 4. Record Data
        if len(population) == 0:
            for _ in range(NUM_ROUNDS - round_num):
                for key in history.keys():
                    history[key].append(0)
            break
            
        strategies = [civ[0] for civ in population]
        wealths = [civ[1] for civ in population]
        
        # Categorize population by type
        t2_civs = []
        t3_civs = []
        
        for civ in population:
            k = K(civ[1])
            if k < 1.0: pass
            elif k < 2.0: pass
            elif k < 3.0: 
                t2_civs.append(civ)
            else: 
                t3_civs.append(civ)
        
        # Overall types count
        history['Type 0'].append(sum(1 for w in wealths if K(w) < 1.0))
        history['Type I'].append(sum(1 for w in wealths if 1.0 <= K(w) < 2.0))
        history['Type II'].append(len(t2_civs))
        history['Type III'].append(len(t3_civs))
        
        # Overall strategies
        history['Communicate'].append(strategies.count('Communicate') / len(population))
        history['Silent'].append(strategies.count('Silent') / len(population))
        history['Attack'].append(strategies.count('Attack') / len(population))
        
        # Type II Strategies
        if len(t2_civs) > 0:
            t2_strats = [c[0] for c in t2_civs]
            history['T2_Comm'].append(t2_strats.count('Communicate') / len(t2_civs))
            history['T2_Sil'].append(t2_strats.count('Silent') / len(t2_civs))
            history['T2_Att'].append(t2_strats.count('Attack') / len(t2_civs))
        else:
            history['T2_Comm'].append(0)
            history['T2_Sil'].append(0)
            history['T2_Att'].append(0)

        # Type III Strategies
        if len(t3_civs) > 0:
            t3_strats = [c[0] for c in t3_civs]
            history['T3_Comm'].append(t3_strats.count('Communicate') / len(t3_civs))
            history['T3_Sil'].append(t3_strats.count('Silent') / len(t3_civs))
            history['T3_Att'].append(t3_strats.count('Attack') / len(t3_civs))
        else:
            history['T3_Comm'].append(0)
            history['T3_Sil'].append(0)
            history['T3_Att'].append(0)

        # ... (end of the for round_num loop) ...

    final_pop_size = len(population)
    return history, final_pop_size

# 5. MONTE CARLO

all_data = {key: [] for key in ['Communicate', 'Silent', 'Attack', 'Type 0', 'Type I', 'Type II', 'Type III', 
                                'T2_Comm', 'T2_Sil', 'T2_Att', 'T3_Comm', 'T3_Sil', 'T3_Att']}

final_populations = []

for trial in range(NUM_TRIALS):
    # Catch the final size here
    hist, final_size = run_simulation()
    final_populations.append(final_size)
    
    for key in all_data.keys():
        all_data[key].append(hist[key])

# Calculate means and stds
means = {key: np.mean(all_data[key], axis=0) for key in all_data.keys()}
stds = {key: np.std(all_data[key], axis=0) for key in all_data.keys()}
rounds = np.arange(NUM_ROUNDS)

# 6. PLOTTING

fig1, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))

# Plot of Strategy Evolution
colors_strat = {'Communicate': 'blue', 'Silent': 'orange', 'Attack': 'green'}
for strat in ['Communicate', 'Silent', 'Attack']:
    ax1.plot(means[strat], label=strat, color=colors_strat[strat], linewidth=2)
    ax1.fill_between(rounds, means[strat] - stds[strat], means[strat] + stds[strat], 
                     color=colors_strat[strat], alpha=0.2)

ax1.set_title('Overall Population: Strategy Evolution')
ax1.set_xlabel('Round')
ax1.set_ylabel('Frequency')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot of Kardashev Type Distribution
colors_type = {'Type 0': 'gray', 'Type I': 'blue', 'Type II': 'orange', 'Type III': 'red'}
for ktype in ['Type 0', 'Type I', 'Type II', 'Type III']:
    ax2.plot(means[ktype], label=ktype, color=colors_type[ktype], linewidth=2)
    ax2.fill_between(rounds, means[ktype] - stds[ktype], means[ktype] + stds[ktype], 
                     color=colors_type[ktype], alpha=0.2)

ax2.set_title('Kardashev Type Distribution')
ax2.set_xlabel('Round')
ax2.set_ylabel('Number of Civilizations')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# 7. PLOTTING TYPE II and III

fig2, (ax3, ax4) = plt.subplots(1, 2, figsize=(18, 6))

# Type II Strategy Evolution
ax3.plot(means['T2_Comm'], label='Communicate', color='blue', linewidth=2)
ax3.fill_between(rounds, means['T2_Comm'] - stds['T2_Comm'], means['T2_Comm'] + stds['T2_Comm'], color='blue', alpha=0.2)

ax3.plot(means['T2_Sil'], label='Silent', color='orange', linewidth=2)
ax3.fill_between(rounds, means['T2_Sil'] - stds['T2_Sil'], means['T2_Sil'] + stds['T2_Sil'], color='orange', alpha=0.2)

ax3.plot(means['T2_Att'], label='Attack', color='green', linewidth=2)
ax3.fill_between(rounds, means['T2_Att'] - stds['T2_Att'], means['T2_Att'] + stds['T2_Att'], color='green', alpha=0.2)

ax3.set_title('Type II Civilizations: Strategy Evolution')
ax3.set_xlabel('Round')
ax3.set_ylabel('Frequency')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Type III Strategy Evolution
ax4.plot(means['T3_Comm'], label='Communicate', color='blue', linewidth=2)
ax4.fill_between(rounds, means['T3_Comm'] - stds['T3_Comm'], means['T3_Comm'] + stds['T3_Comm'], color='blue', alpha=0.2)

ax4.plot(means['T3_Sil'], label='Silent', color='orange', linewidth=2)
ax4.fill_between(rounds, means['T3_Sil'] - stds['T3_Sil'], means['T3_Sil'] + stds['T3_Sil'], color='orange', alpha=0.2)

ax4.plot(means['T3_Att'], label='Attack', color='green', linewidth=2)
ax4.fill_between(rounds, means['T3_Att'] - stds['T3_Att'], means['T3_Att'] + stds['T3_Att'], color='green', alpha=0.2)

ax4.set_title('Type III Civilizations: Strategy Evolution')
ax4.set_xlabel('Round')
ax4.set_ylabel('Frequency')
ax4.legend()
ax4.grid(True, alpha=0.3)

print()
plt.tight_layout()
plt.show()

# 9. PRINTING THE FINAL SURVIVAL TABLE

final_t0 = [trial[-1] for trial in all_data['Type 0']]
final_t1 = [trial[-1] for trial in all_data['Type I']]
final_t2 = [trial[-1] for trial in all_data['Type II']]
final_t3 = [trial[-1] for trial in all_data['Type III']]

final_total = [t0 + t1 + t2 + t3 for t0, t1, t2, t3 in zip(final_t0, final_t1, final_t2, final_t3)]

print("\n" + "="*65)
print(f"FINAL SURVIVAL TABLE (Average of {NUM_TRIALS} Monte Carlo Trials)")
print("="*65)
print(f"{'Kardashev Type':<18} | {'Mean Survivors':<15} | {'Std Dev':<10} | {'Max Survivors'}")
print("-" * 65)

# Print a row for each type
print(f"{'Type 0 (Planetary)':<18} | {np.mean(final_t0):>14.1f} | {np.std(final_t0):>10.1f} | {np.max(final_t0):>10}")
print(f"{'Type I (Planetary)':<18} | {np.mean(final_t1):>14.1f} | {np.std(final_t1):>10.1f} | {np.max(final_t1):>10}")
print(f"{'Type II (Stellar)':<18} | {np.mean(final_t2):>14.1f} | {np.std(final_t2):>10.1f} | {np.max(final_t2):>10}")
print(f"{'Type III (Galactic)':<18} | {np.mean(final_t3):>14.1f} | {np.std(final_t3):>10.1f} | {np.max(final_t3):>10}")

print("-" * 65)
print(f"{'TOTAL SURVIVORS':<18} | {np.mean(final_total):>14.1f} | {np.std(final_total):>10.1f} | {np.max(final_total):>10}")
print("="*65 + "\n")
````

The figures above show a Monte Carlo simulation of 1000 civilisations over 100 rounds with the aim to model technological growth, interaction between civilisations of varying degrees of advancement, and asymmetric warfare on a cosmic scale.

This model starts by creating 1000 civilisations with a randomly chosen strategy. Each round, two civilisations are paired up at random. Both have a chance to discover one another with probability $q$. If one discovers the other, they are either attack, stay silent, or initiate communication according to their strategy type. If both discover the other, the model randomly chooses one with a 50% probability to make the first move. This is repeated for 100 _rounds_. If a civilisation's $R$ drops to zero or less, it is considered dead. Additionally, infant civilisations (Type 0) have a chance of being born each round with probability $0.5$.


### **Analysing the results**

#### The Early Game (Round 0-20)
Though all civilisation types start out with equal numbers, the _Kardashev Type Distribution_ plot shows a massive spike in Type I civilizations, accompanied by a sharp decrease in the number of Type 0 civilisations. This is because over a few rounds, the technological explosion of the Type 0 civilisations advances them toward Type I civilisations. However, the number of Type I civilisations sharply decrease, as their probability of detection increases. The aggressive advanced civilisations quickly pick off any discovered Type I civilisations. The number of Type III civilisations also decreases. Attacks between Type III civilisations are very powerful, and could drop them down the attacked to Type II civilisations. Furthermore, surviving Type I civilisations advance to Type II. This causes the number of Type II civilisations to stay more stable than the others. 


Because their strike probability is near zero, Type 0 and Type I civilizations that choose the Attack strategy are rapidly eliminated by retaliation. The _Strategy Evolution_ plot shows a distinct dip in _Attack_ during this phase. 

#### The Mid Game (Round 20 - 60)
Type IIs are the most abundant civilization in our simulated cosmos during this era. In this stage, the Type IIs are predominantly silent. They are strong enough to effortlessly attack weaker civilizations, but they remain highly vulnerable to a peer or a Type III. Because the value of cooperation ($V$) is too low to justify the catastrophic risk of betrayal, Communication is driven to extinction. The galaxy settles into silence, as the expected damage of war vastly outweighs the potential rewards of trade.

#### The Late Game (Round 60-99)
As the surviving civilizations climb to Type III status, the Dark Forest ignites. The Type IIIs initially annihilate each other in mutual interstellar wars. However, as these destructive encounters become too common, Silence emerges as the safer strategy among the apex predators. But when Silence amongst the Type IIs and IIIs becomes too common, the remaining Attackers can freely destroy the hidden ones they manage to find. Consequently, pre-emptive strikes surge once again, and the Attack strategy begins its ascend. 

#### The Great Silence (Round 100+)
The surge in pre-emptive strikes leads to the ultimate tragedy of the Dark Forest. The Type IIIs, the endless wars and the inability to regrow their lost resources slowly bleed them dry. One by one, the predators bleed out and are demoted to Type IIs, subsequently instantly annihilated by the remaining Type IIIs. The simulation concludes with near-total extinction, leaving behind a silent, empty galaxy.

### Limitations
1. **The Absence of Pre-emptive Strike Logic**

As discussed in the section _Pre-emptive Strikes_ A Type III civilization strikes first not just because it is hostile, but because it knows that if it waits, its peer will eventually find and destroy it. However, in the simulation, interactions are hard-coded, and a civilisation does not adjust their strategy according to what may best suit them. It does not model the civilisation's strategy, that being striking first to gain a first-mover advantage. Because the simulation lacks pre-emptive logic, it may actually underestimate the frequency of attacks among Type III civilizations. It is possible that the chain of suspicion would likely drive the Attack strategy to an even higher frequency than observed here.

2. **The Automatic Retaliation Assumption**

The simulation assumes that if Civilization A attacks Civilization B and misses, B will automatically retaliate (assuming B survives). In reality, a civilization might choose not to retaliate out of fear, or it might lack the capability to strike back. However, by forcing automatic retaliation, we give the Dark Forest its strongest possible case: war is always punished. Relaxing this assumption (allowing defenders to remain passive) would only make attacking safer, likely accelerating the self-terminating extinction of the galaxy.

3. **Lack of Spacial Geography**

Our model assumes a well-mixed population where any civilization can interact with any other. In reality, the vastness of space creates geographic isolation. Future iterations should introduce a spatial grid or network topology, allowing localized clusters of civilizations to evolve independently.

4. **Lack of Paranoia**

The simulation assumes perfect honesty in signalling. We do not model the possibility of deception, where a malicious civilization broadcasts a cooperative message to lure a hidden target out of silence. Adding a 'deception' parameter—where communication can be a trap—would further darken the forest and likely accelerate the collapse.

5. **Subject to Parameters**

The computational results are highly sensitive to the chosen scaling constants, particularly the value of cooperation ($V$). While we calibrated these constants to reflect human technological history and to give the Dark Forest its strongest case, they remain theoretical. 

6. **Unified Civilisations**

We model civilizations as unitary actors with a single, cohesive strategy. In reality, interstellar empires are likely to be politically heterogeneous. One faction might launch a rogue preemptive strike, dragging the entire empire into war. Modeling internal factional dynamics is a promising avenue for future research.

7. **The Cosmic Speed Limit**

Our model assumes instantaneous interaction. However, interstellar communication and warfare are bounded by the speed of light. A strike launched at a Type 0 civilization might arrive millions of years later, finding a Type III civilization instead. Modeling these temporal delays would add a crucial layer of chaotic unpredictability to the Dark Forest.

8. **Interactions as Duels**

Our computational simulation models all interactions as isolated, pairwise duels between two civilizations. In a crowded galaxy, encounters may involve multiple civilizations simultaneously. This means the simulation cannot capture the formation of alliances or coalitions. Furthermore, the model misses the 'bystander effect,' where a third civilization watches two others exhaust each other in a war and then swoops in to destroy the weakened victor. Introducing multi-agent interactions would fundamentally alter the evolutionary dynamics, potentially allowing weaker civilizations to survive by forming defensive pacts against the apex predators."

9. **Total Isolation/Perfect Hiding**

Our model assumes that civilizations must remain active participants in the galactic ecosystem. However, Liu Cixin's concept of the "Black Domain" (from _Death's End_) offers an alternative: a civilization can alter the physical laws of their local space to create a region of absolute isolation. They cannot be attacked, they cannot be betrayed. However this comes at a cost being that they will be isolated in that this domain for the rest of time. Perhaps advanced civilisations find a way to transform all their matter into dark matter, leaving them invisible. We can only speculate as to the limits of their technological capabilities.

# 4. Conclusion
Our analytical model revealed that the Dark Forest holds across the Kardashev scale. For Type I civilizations, the Dark Forest is a state of forced silence. Lacking the offensive capability to strike, they are mathematically forced into the role of prey. For Type III civilizations, the Dark Forest manifests as active predation, driven by the chain of suspicion and the catastrophic cost of being wrong.

The tragedy of the Dark Forest is revealed when we map it to the classic Prisoner’s Dilemma. In this cosmic game, Silence represents cooperation, and Attack represents defection. Communication is a naive strategy that is eliminated by the evolutionary filter early on. The rest of the game is played between the hunters and the hidden prey. 

In the middle to late game, when two Type III civilizations meet, they face the classic dilemma. If both choose Silence (Mutual Cooperation), they survive and continue to grow. If both choose Attack (Mutual Defection), they enter an endless exchange of blows that devastates both. However, because the _Sucker's payoff_ (choosing Silence while the other chooses Attack) is an immense loss or extinction, the fear of exploitation forces both rational actors to defect.

In a standard Prisoner's Dilemma, the Nash Equilibrium is mutual defection. Our computational simulation proved that the Dark Forest is the evolutionary realization of this Nash Equilibrium. When the value of cooperation ($V$) is low, the incentive to defect becomes overwhelming. The Attack strategy dominates, and the chain of suspicion forces the apex predators into war. 

However, we also revealed a twist to the game. In the standard Prisoner's Dilemma, mutual defection results in a suboptimal but survivable outcome. But in the Dark Forest, the stakes are absolute. Mutual defection does not just lead to a bad outcome; it leads to only to mutual destruction. The Type III civilizations are locked into an endless wars. They lose their resources faster than their growth could recover. The predators annihilated each other, demoting themselves into prey, until the galaxy was entirely empty. This is The Great Silence. We do not hear from other civilizations because the rational pursuit of survival in a universe of infinite stakes inevitably leads to mutual destruction. 

The Dark Forest is The Prisoner's Dilemma of the cosmos. As long as the risk of being wrong is fatal, no one can afford to speak.
